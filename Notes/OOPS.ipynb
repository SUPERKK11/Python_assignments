## **Constructor Chainning** 

```python
Calling or invoking the constructor of parent class inside the constructor of child class that is known as constructor chainning.
Three ways for the Constructor Channing :
1. super().__init__(args) # this is used in latest version
2. super(child_class_name,self).__init__(args) # this was used in older versions of python 
3. parent_class_name.__init__(self,args)
```

## **Method chainning**

```python

Method chainning calling or invoking the method of parent class inside the method of child class that is known as method chainning.
there are three ways for the method chainning:

1. super().method_name(args)
2. super(child_class_name,self/cls).method_name(args)
3. parent_class_name.method_name(self/cls,args)

```

```python

Method chaining is a programming technique where you call multiple methods sequentially on a single object in a single line of code.

Example:-

            class DataProcessor:
                def __init__(self, data):
                    self.data = data
                def remove_duplicates(self):
                    # Logic to remove duplicates (simulated here)
                    self.data = list(set(self.data))
                    return self  # <--- Crucial step for chaining
                def sort_data(self):
                    self.data.sort()
                    return self  # <--- Crucial step for chaining
                def scale_values(self, factor):
                    self.data = [x * factor for x in self.data]
                    return self  # <--- Crucial step for chaining
                def get_result(self):
                    # Terminal method: returns the final data, not 'self'
                    return self.data
            # Usage: Chaining methods in a single, readable pipeline
            raw_data = [3, 1, 2, 3, 5, 2]
            processor = DataProcessor(raw_data)
            # The Method Chain
            clean_data = processor.remove_duplicates().sort_data().scale_values(10).get_result()
            print(clean_data) 
            # Output: [10, 20, 30, 50]
            
```

````python
class DataProcessor:
    def __init__(self, data):
        self.data = data
    def remove_duplicates(self):
        # Logic to remove duplicates (simulated here)
        self.data = list(set(self.data))
        return self  # <--- Crucial step for chaining
    def sort_data(self):
        self.data.sort()
        return self  # <--- Crucial step for chaining
    def scale_values(self, factor):
        self.data = [x * factor for x in self.data]
        return self  # <--- Crucial step for chaining
    def get_result(self):
        # Terminal method: returns the final data, not 'self'
        return self.data
# Usage: Chaining methods in a single, readable pipeline
raw_data = [3, 1, 2, 3,2,5,4]
processor = DataProcessor(raw_data)
# The Method Chain
clean_data = processor.remove_duplicates().sort_data().scale_values(10).get_result()
print(clean_data) 
# Output: [10, 20, 30, 50]
````

## **3.Multilevel Inheritance**

```python

In multilevel inheritance, features of the base class and the derived class are further inherited into the new derived class. This is similar to a relationship representing a child and a grandfather. 

```

```python

                # Base class
                class Grandfather:
                    def __init__(self, grandfathername):
                        self.grandfathername = grandfathername

                # Intermediate class
                class Father(Grandfather):
                    def __init__(self, fathername, grandfathername):
                        self.fathername = fathername
                        # Call the constructor of Grandfather
                        Grandfather.__init__(self, grandfathername)

                # Derived class
                class Son(Father):
                    def __init__(self, sonname, fathername, grandfathername):
                        self.sonname = sonname
                        # Call the constructor of Father
                        Father.__init__(self, fathername, grandfathername)

                    def print_name(self):
                        print('Grandfather name :', self.grandfathername)
                        print('Father name :', self.fathername)
                        print('Son name :', self.sonname)

                # Driver code
                s1 = Son('Prince', 'Rampal', 'Lal mani')
                print(s1.grandfathername)
                s1.print_name()
                    
```

## **METHOD OVERLOADING**

### **VVV.IMP** WE cannot achieve method overloading in python
```python
Unlike languages like Java or C++, Python does not support traditional method overloading natively.

If you define multiple methods with the same name in a single Python class, the last method defined simply overwrites the previous ones. However, because Python is dynamically typed and highly flexible, you can achieve the exact same behavior using a few different "Pythonic" techniques.

```

## **Here are the standard ways to replicate method overloading in Python:**

```python
1. Using Default Arguments (The Simplest Way)
If you want a method to handle a varying number of parameters, you can assign default values to the optional ones.
eg:-
class Calculator:
    def add(self, a, b, c=0):
        # If 'c' isn't provided, it defaults to 0
        return a + b + c

calc = Calculator()
print(calc.add(2, 3))    # Output: 5 (uses default c=0)
print(calc.add(2, 3, 4)) # Output: 9
```


```python
2. Using *args and **kwargs (The Flexible Way)
"If you dont know how many arguments will be passed, you can use *args (for a variable number of positional arguments) or **kwargs (for keyword arguments)."
eg:-
class Calculator:
    def add(self, *args):
        # args is a tuple of all passed arguments
        total = 0
        for num in args:
            total += num
        return total

calc = Calculator()
print(calc.add(5, 10))        # Output: 15
print(calc.add(1, 2, 3, 4))   # Output: 10
```

````python
from abc import ABC,abstractmethod

class Cal(ABC):
    
    @abstractmethod
    def add(a,b):
        pass
    
    @abstractmethod
    def sub(a,b):
        pass
        
        
class Update_Cal(Cal):
    def add(self,a,b):
        print(a+b)
    def sub(self,a,b):
        print(a-b)
        
ob1=Update_Cal()
ob1.add(1,3)
````
