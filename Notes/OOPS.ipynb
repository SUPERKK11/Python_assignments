## **Constructor Chainning** 

```python
Calling or invoking the constructor of parent class inside the constructor of child class that is known as constructor chainning.
Three ways for the Constructor Channing :
1. super().__init__(args)
2. super(child_class_name,self).__init__(args)
3. parent_class_name.__init__(self,args)
```

## **example(super().__init__(args))**

````python
class university:
    Uname="IIT"
    Uloc="Mumbai"
    Umail="HCverma@gmail.com"
    def __init__(self,name,age):
        self.name=name
        self.age=age
class student(university):
    def __init__(self,name,age,SID,SLOC):
        super().__init__(name,age)
        self.SID=SID
        self.SLOC=SLOC
    def display(self):
        print(self.name,self.age,self.SID,self.SLOC)
s1=student("A",18,919,"pune")
s1.display()
    
````

## **Method chainning**

```python
Method chainning calling or invoking the method of parent class inside the method of child class that is known as method chainning.
there are three ways for the method chainning:
1. super().method_name(args)
2. super(child_class_name,self/cls).method_name(args)
3. parent_class_name.method_name(self/cls,args)
```

```python
Method chaining is a programming technique where you call multiple methods sequentially on a single object in a single line of code.
Example:-
class DataProcessor:
    def __init__(self, data):
        self.data = data
    def remove_duplicates(self):
        # Logic to remove duplicates (simulated here)
        self.data = list(set(self.data))
        return self  # <--- Crucial step for chaining
    def sort_data(self):
        self.data.sort()
        return self  # <--- Crucial step for chaining
    def scale_values(self, factor):
        self.data = [x * factor for x in self.data]
        return self  # <--- Crucial step for chaining
    def get_result(self):
        # Terminal method: returns the final data, not 'self'
        return self.data
# Usage: Chaining methods in a single, readable pipeline
raw_data = [3, 1, 2, 3, 5, 2]
processor = DataProcessor(raw_data)
# The Method Chain
clean_data = processor.remove_duplicates().sort_data().scale_values(10).get_result()
print(clean_data) 
# Output: [10, 20, 30, 50]
```

````python
class DataProcessor:
    def __init__(self, data):
        self.data = data
    def remove_duplicates(self):
        # Logic to remove duplicates (simulated here)
        self.data = list(set(self.data))
        return self  # <--- Crucial step for chaining
    def sort_data(self):
        self.data.sort()
        return self  # <--- Crucial step for chaining
    def scale_values(self, factor):
        self.data = [x * factor for x in self.data]
        return self  # <--- Crucial step for chaining
    def get_result(self):
        # Terminal method: returns the final data, not 'self'
        return self.data
# Usage: Chaining methods in a single, readable pipeline
raw_data = [3, 1, 2, 3,2,5,4]
processor = DataProcessor(raw_data)
# The Method Chain
clean_data = processor.remove_duplicates().sort_data().scale_values(10).get_result()
print(clean_data) 
# Output: [10, 20, 30, 50]
````

## **METHOD OVERLOADING**

### **VVV.IMP** WE cannot achieve method overloading in python
```python
Unlike languages like Java or C++, Python does not support traditional method overloading natively.

If you define multiple methods with the same name in a single Python class, the last method defined simply overwrites the previous ones. However, because Python is dynamically typed and highly flexible, you can achieve the exact same behavior using a few different "Pythonic" techniques.

```

## **Here are the standard ways to replicate method overloading in Python:**

```python
1. Using Default Arguments (The Simplest Way)
If you want a method to handle a varying number of parameters, you can assign default values to the optional ones.
eg:-
class Calculator:
    def add(self, a, b, c=0):
        # If 'c' isn't provided, it defaults to 0
        return a + b + c

calc = Calculator()
print(calc.add(2, 3))    # Output: 5 (uses default c=0)
print(calc.add(2, 3, 4)) # Output: 9
```


```python
2. Using *args and **kwargs (The Flexible Way)
"If you dont know how many arguments will be passed, you can use *args (for a variable number of positional arguments) or **kwargs (for keyword arguments)."
eg:-
class Calculator:
    def add(self, *args):
        # args is a tuple of all passed arguments
        total = 0
        for num in args:
            total += num
        return total

calc = Calculator()
print(calc.add(5, 10))        # Output: 15
print(calc.add(1, 2, 3, 4))   # Output: 10
```

## **OPERATOR OVERLOADING**


```python
Same operator performs more than one task is known as operator overloading .
- All the operators can perform operations on the inbuilt object but none of the operator can perform operation on user define object.

- To perform operation on the user define object through we have to introduce magic method

- Magic method :- magic method is also known as dunder method (because it starts of double underscore)
this method is called through operator . 

each and every operator have thier own magic method 

for example:-
+ -> __add__()
- -> __sub__()
```

````python
class A:
    def __init__(self,m):
        self.m=m
    def __add__(self,other):
        return A(self.m+other.m)
ob1=A(10)
ob2=A(20)
ob1+ob2
````

## Python __repr__() magic method

```python
Python __repr__() is one of the magic methods that returns a printable representation of an object in Python that can be customized or predefined, i.e. we can also create the string representation of the object according to our needs.
Python __repr__() magic method Syntax

Syntax: object.__repr__()

object: The object whose printable representation is to be returned.

Return: Simple string representation of the passed object
```

````python
class A:
    def __init__(self,m):
        self.m=m
    def __add__(self,other):
        return A(self.m+other.m) # this returns the address  #<__main__.A at 0x10d4f75c0>
    def __repr__(self):
        return str(self.m) # this only resturns string 
ob1=A(10)
ob2=A(20)
ob3=A(30)
ob2+ob1+ob3
````
