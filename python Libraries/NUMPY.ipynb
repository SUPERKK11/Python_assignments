````python
#RECURRSION

def flatten(arr):
    result = []
    for x in arr:
        if isinstance(x, list):
            result.extend(flatten(x))
        else:
            result.append(x)
    return result


arr = [[[1, 2, 3, 4], [1, 1, 34, 2]]]

print(flatten(arr))
````
