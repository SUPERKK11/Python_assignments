````python
#RECURRSION

def flatten(arr):
    result = []
    for x in arr:
        if isinstance(x, list):
            result.extend(flatten(x))
        else:
            result.append(x)
    return result


arr = [[[1, 2, 3, 4], [1, 1, 34, 2]]]

print(flatten(arr))
````

## **FANCY Indexing** <hr>

```python 
Fancy indexing in NumPy refers to accessing or modifying multiple array elements at once by passing lists or arrays of integer indices. Unlike basic slicing (like arr[1:5]), fancy indexing allows you to extract elements that are non-adjacent, scattered, or completely out of order
```
