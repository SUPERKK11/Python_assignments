````python
def armstrong_number(n):
    def count_n(n,count=0):
        if n == 0:
            return count
        digit=n%10
        count=count+1
        return count_n(n//10,count)
    def summation(n,sum=0,count=0):
        if n == 0:
            return sum
        if count == 0:
            count=count_n(n)
        digit=n%10
        sum=sum+digit**count
        return summation(n//10,sum,count)
    if n == summation(n):
        return True
    else:
        return False
armstrong_number(153)
    
    
````

````python
def largest_number(arr):
    curr_max=arr[0]
    for i in arr:
        if i>curr_max:
            curr_max=i
    return curr_max
def second_largest_number(arr):
    curr_max=None
    if arr[0] != largest_number(arr):
        curr_max=arr[0]
    else:
        curr_max=arr[1]
    for i in arr:
        if i<largest_number(arr) and i>curr_max:
            curr_max=i
    return curr_max
second_largest_number([2,25,8])
````

````python
def remove_duplicates(arr):
    s=set()
    count=0
    for i in range(len(arr)):
        if arr[i] not in s:
            s.add(arr[i])
            arr[count]=arr[i]
            count=count+1
    return arr[:count]
remove_duplicates([2,1,2,3,8,8])
    
        
        
````

````python
def rotate_right(arr,k):
    k=k%(len(arr))
    arr[:]=arr[::-1][:k][::-1]+arr[::-1][k:][::-1]
    return arr
rotate_right([1,2,3,4],1)
````

````python
def gcd(a,b):
    if a>b:
        greater_num=a
        smaller_num=b
    else:
        greater_num=b
        smaller_num=a
    while smaller_num != 0:
        greater_num,smaller_num=smaller_num,greater_num%smaller_num
    return greater_num
gcd(8,12)
````

````python
def prime(n):
    if n == 2:
        return False
    for i in range(2,n):
        if n%i == 0:
            return True
        else:
            return False            
def prime_n(n1,n2):
    for i in range(n1,n2+1):
        prime(i)
        if prime(i) is not None and prime(i) is False:
            print(i)
prime_n(3,10)
````
