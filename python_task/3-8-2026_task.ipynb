```python
n=int(input())
for row in range(n):     
    for col in range(n):
        if row>=col:
            if row == 0:
                print(1,end=" ")
            if row == 1 and col == 0:
                print(6,end=" ") 
            if row == 1 and col == 1:
                print(4,end=" ")
            if row == 2 :
                print(row+6,end=" ") 
            if row == 3 :
                print(row+13,end=" ")
        else:
            print(" ", end=" ")
    print()
        
```
