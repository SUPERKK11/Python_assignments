```text
1.Write a program to count palindrome and disarium numbers in an array.
```

````python
def reverse(i,rev=0):
    if i == 0:
        return 
    digit=i%10
    rev=rev+digit
    reverse(i//10)
    return rev
def sum_squared_digit(i,s_s_d=0):
    if i == 0:
        return
    digit=i%10
    s_s_d=s_s_d+digit**2
    sum_squared_digit(i//10)
    return s_s_d
def disarium(i):
    if i == sum_squared_digit(i):
        return True   
def palindrome(i):
    if i == reverse(i):
        return True
def count_palindrome_disarium(arr,count=0):
    for i in arr:
        if palindrome(i) and disarium(i):
            count=count+1
    return count
arr=[1,2,3,4]
count_palindrome_disarium(arr)
    
````

```text
2.Write a program to search for a given element in an array and display its index.
```

````python
def binary_search(t,arr):
    i=0
    j=len(arr)-1
    while i<=j:
        mid=(i+j)//2
        if arr[mid] == t:
            return mid
        elif arr[mid]<t:
            i=mid+1
        else:
            j=mid-1
arr=[1,2,3,4,5]
result=binary_search(5,arr)
if result:
    print(result)
else:
    print("element is not in arr")
    
````

```text
3.Write a program to reverse an array without using another array.
```

````python
def reverse(arr,temp=0):
    i=0
    j=len(arr)-1
    while i <=j:
        temp=arr[i]
        arr[i]= arr[j]
        arr[j]=temp
        i=i+1
        j=j-1
    return arr
arr=[1,2,3,4]
reverse(arr)

        
````

```text
4.Write a program to find the second smallest element in an array.
```

````python
def smallest(arr,small=None):
    small=arr[0]
    for i in range(len(arr)):
        if arr[i]< small:
            small=arr[i]
    return small
def second_smallest(arr,s_small=None):
    small=smallest(arr)
    s_small=arr[0]
    if small == s_small:
        return s_small
    for i in range(len(arr)):
        if s_small is not None:
            if arr[i] > small and arr[i] < s_small:
                s_small=arr[i]
    return s_small
arr=[5,2,4]
second_smallest(arr)
    
````

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n*factorial(n-1)
factorial(5)
````

````python
def factorial(n,i=1,fact=1):
    if i == n+1:
        return fact
    fact=fact*i
    return factorial(n,i+1,fact)
factorial(5)
````
