```python
#palindrome using two - pointers

def is_palindrome(s):
    start=0
    end=len(s)-1
    while start<end:
        if s[start] != s[end]:
            print("not a palindrome")
            break
        start=start+1
        end=end-1
    else:
        print("palindrome")
        
is_palindrome("mom")
    

    
```

```python
#reverse the list
def reverse_list(list1):
    start=0
    end=len(list1)-1
    while start < end:
        list1[start],list1[end]=list1[end],list1[start]
        start=start+1
        end=end-1
    return list1
reverse_list([1,2,3,4])
```

```python
#OOP programming question
class Mobile:
    def __init__(self,brand,price):
        self.brand=brand
        self.price=price
    def display(self):
        colours=["red","blue","green"]
        for i in colours:
            print(self.brand,self.price,i)
ob1=Mobile("apple",100000)
ob1.display()
        
```
