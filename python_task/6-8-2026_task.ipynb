````python
def iseven(n):
    if n==0:
        return False
    if n%2 == 0:
        return True
    else:
        return False
iseven(0)
````

````python
def reverse(str1):
    str2=""
    for i in range(len(str1)-1,-1,-1):
        str2=str2+str1[i]
    return str2
reverse("hello")
````

````python
def find_max(arr):
    current_max=arr[0]
    for i in arr:
        if i>current_max:
            current_max=i
    return current_max
find_max([1,2,3,4])
````

````python
def reverse_star(n):
    for col in range(n):
        for row in range(n):
            if row>col:
                print("*",end=" ")
        print()
reverse_star(4)
    
````

````python
def pattern(n):
    for row in range(1,n+1):
        for col in range(1,n+1):
            if row>=col:
                if row == 1 and col == 1:
                    print(1,end=" ")
                elif row == 2 and col == 1:
                    print(6,end=" ")
                elif row == 2 and col == 2:
                    print(4,end=" ")
                elif row == 3 and col == 1:
                    print(8,end=" ")
                elif row == 3 and col == 2:
                    print(9,end=" ")
                elif row == 3 and col == 3:
                    print(10,end=" ")
                elif row == 4 and col == 1:
                    print(16,end=" ")
                elif row == 4 and col == 2:
                    print(15,end=" ")
                elif row == 4 and col == 3:
                    print(14,end=" ")
                elif row == 4 and col == 4:
                    print(12,end=" ")
                elif row == 5 and col == 1:
                    print(18,end=" ")
                elif row == 5 and col == 2:
                    print(20,end=" ")
                elif row == 5 and col == 3:
                    print(21,end=" ")
                elif row == 5 and col == 4:
                    print(22,end=" ")
                elif row == 5 and col == 5:
                    print(24,end=" ")                 
        print()
pattern(5)
````

````python
def is_non_prime(num): 
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count += 1     
    # Non-primes have a divisor count that is NOT 2
    return count != 2

def print_pattern(n_rows):
    current_num = 1
    for row in range(1, n_rows + 1):
        temp_list = []
        # Keep finding non-primes until we have 'row' amount of numbers
        while len(temp_list) < row:
            if is_non_prime(current_num):
                temp_list.append(current_num)
            current_num += 1
        # If the row is even (2, 4, 6...), reverse the list
        if row % 2 == 0:
            temp_list.reverse()    
        # Convert the numbers to strings and join them with spaces
        print(" ".join(map(str, temp_list)))

if __name__ == "__main__":
    print_pattern(6)
````
