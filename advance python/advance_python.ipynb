````python
#decorator

def decorator(func):
    def wrapper(*args,**kwargs):
        print("hello")
        func(*args,**kwargs)
        print("bye")
    return wrapper
    
# @decorator
# def A():
#     print("I am K")
````

````python

def A():
    print("I am K")
#or
x=decorator(A)
x()
````

````python

def delay(func):
    def inner_delay(*args,**kwargs):
        import time 
        time.sleep(5)
        func(*args,**kwargs)
    return inner_delay

@delay
def A():
    print("hello")
A()


````

````python

def delay(func):
    def inner_delay(*args,**kwargs):
        
        import time 
        start=time.time()
        func(*args,**kwargs)
        end=time.time()
        print("time taken to execute the function is :",end-start)
    
    return inner_delay

@delay
def A():
    print("hello")
A()
````

````python
def n_1_20():
    for i in range(1,21):
        yield i
list(n_1_20())
````

````python
def fibonacci(n):
    a=0
    b=1
    for i in range(1,n+1):
        yield a
        a,b=b,b+a
list(fibonacci(12))
    
````

````python
def h():
    for i in range(ord("a"),ord("z")):
        yield chr(i),i
dict(h())
````
