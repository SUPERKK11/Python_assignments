
````python
class School:
    """ these are class member """
    s_name="DAV"
    S_Loc="Pak"
    S_mail="DAV@pak.com"
    S_timing="4am"
St1=School()
""" these are object member """
St1.name="A"
St1.mail="A@gmail.com"

St2=School()
St2.name="B"
St2.mail="B@gmail.com"

'''
we can access all the class member by st1 and st2 because they are common for all the object
and also we access their own property that is name and mail (example from ST1)
'''


````

```python
2. Class method :- The method that is used to access and modify the class member that method is known as class method
here , class method , decorator and cls are used 
syntax:-
class C_name:
    code
    @classmethod #decorator
    def method_name(cls,args):
        code
obj_name=Cname(value)
obj_name.method_name(value)
or 
C_name.method_name(obj_name,value)
eg:-
class Bank:
    bname="SBI"
    bloc="pak"
    bmanager="Hamaza ali"
    @classmethod
    def display(cls):
        print(cls.bname,cls.bloc,cls.bmanager)

obj_name=Bank()
obj_name.display()
or 
Bank.display(obj_name)

```

````python
class Bank:
    bname="SBI"
    bloc="pak"
    bmanager="Hamaza ali"
    @classmethod
    def display(cls):
        print(cls.bname,cls.bloc,cls.bmanager)
    @classmethod
    def change_loc(cls,new_loc):
        cls.bloc=new_loc
        
obj_name=Bank()
obj_name.display()
obj_name.change_loc("India")
# or 
Bank.display()
````
