### Types of states/properties/members

```python
type of states are classified into two types 
                             |
--------------------------------------------------------------
|                                                            |
1.class member / generic member                   2.Object member / specific member
```

## Class member

```python
The member that is common for each and every object that we created 
that is known as class member

```

## Object Member

```python
The member that is different for each and every object that 
is known as object member
```

# example:

````python
class School:
    #these are class member
    s_name="DAV"
    S_Loc="Pak"
    S_mail="DAV@pak.com"
    S_timing="4am"
St1=School()
#these are object member
St1.name="A"
St1.mail="A@gmail.com"

St2=School()
St2.name="B"
St2.mail="B@gmail.com"

#we can access all the class member by st1 and st2 because they are common for all the object

````
